```python
mental_health_df["What is your CGPA?"].unique()
mental_health_df["What is your CGPA?"].str.strip()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=mental_health_df, x="What is your CGPA?",color='red')
plt.title("Mental health histogram")

plt.show()
```

Exercise 5

```python
sns.countplot(data=mental_health_df, x='Choose your gender', hue='Do you have Anxiety?', palette={"Yes":"green","No":"red"})
plt.title("Gender Bar graph")
plt.show()

```

Exercise 5 another way

```python
mental_health_df["Anxiety_encoded"] = mental_health_df["Do you have Anxiety?"].map({"Yes": 1, "No": 0})
```

```python
mental_health_df["Anxiety_encoded"].unique()
```

```python
sns.barplot(data=mental_health_df, x='Choose your gender', y="Anxiety_encoded")
plt.title("Gender Bar graph")
plt.show()
```

```python
mental_health_df['Do you have Panic attack?'].unique()
```

```python
mental_health_df["Panic_encoded"] = mental_health_df['Do you have Panic attack?'].map({"Yes": 1, "No": 0})
```

```python
sns.scatterplot(data=mental_health_df, x='Age', y='Panic_encoded', alpha=0.5, s=80, color="blue", marker="o")
plt.title("Age panic attack")
plt.xlabel("Age")
plt.ylabel("Panic Attacks (0 = No, 1 = Yes)")
plt.show()
```
