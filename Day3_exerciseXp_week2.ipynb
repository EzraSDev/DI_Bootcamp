```python
import numpy as np

data = [12, 15, 13, 12, 18, 20, 22, 21]
```

```python
#Exercise 2

mean_np = np.mean(data)
meadian_np = np.median(data)
variance_np = np.var(data)
std_dev_np = np.std(data)

print(mean_np)
print(meadian_np)
print(variance_np)
print(std_dev_np)
```

```python
#exercise 3

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm


x = np.linspace(20, 80, 100)
y = norm.pdf(x, loc=50, scale=10)

plt.plot(x, y)
plt.show
```

```python
# Exercise 4

from scipy import stats

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_statistic, p_value = stats.ttest_ind(data1, data2)
print(t_statistic)
print(p_value)
```

```python
# Exercise 5

from sklearn.linear_model import LinearRegression

house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150.000, 200.000, 210.000, 250.000, 280.000])

model = LinearRegression()
model.fit(house_sizes.reshape(-1,1), house_prices)

print("Slope:", model.coef_)
print("Intercept", model.intercept_)
```

```python
prediction = model.predict([[90]])
print("Prediction:", prediction)
```

```python
# Exercise 6

fertilizer_1 = np.array([5, 6, 7, 6, 5])
fertilizer_2 = np.array([7, 8, 7, 9, 8])
fertilizer_3 = np.array([4, 5, 4, 3, 4])

f_statistic, p_value = stats.f_oneway(
    fertilizer_1, fertilizer_2, fertilizer_3
)
print("F-value:", f_statistic)
print("P- Value:", p_value)

# We reject the null hypothesis, that means the fertilier is effective and all of them have the average effect on the growth
# but if it was higher than 0.05 we fail to reject it, meaning that we cant conclude the fertilizer didnt help
```
