```python
aboard_check = (
    accidents_df['Aboard Passangers'] +
    accidents_df['Aboard Crew']
)

fatalities_check = (
    accidents_df['Fatalities Passangers'] +
    accidents_df['Fatalities Crew']
)

print("Aboard mismatches:",
      (accidents_df['Aboard'] != aboard_check).sum())

print("Fatalities mismatches:",
      (accidents_df['Fatalities'] != fatalities_check).sum())
```

```python
accidents_df['Sum of fatalities'] = accidents_df['Fatalities Crew'] + accidents_df['Fatalities Passangers']
```

```python
diff = accidents_df['Fatalities'].compare(accidents_df['Sum of fatalities'])
print(diff)
```

```python
accidents_df['Fatalities'] = accidents_df['Fatalities'].fillna(accidents_df['Sum of fatalities'])
```

```python
accidents_df['Fatalities'].isna().sum()
```

```python
missing_passengers = (
    accidents_df['Fatalities Passangers'].isna() &
    accidents_df['Fatalities'].notna() &
    accidents_df['Fatalities Crew'].notna()
)

missing_crew = (
    accidents_df['Fatalities Crew'].isna() &
    accidents_df['Fatalities'].notna() &
    accidents_df['Fatalities Passangers'].notna()
)

print("Recoverable passenger values:", missing_passengers.sum())
print("Recoverable crew values:", missing_crew.sum())
```

```python
accidents_df.loc[missing_passengers, 'Fatalities Passangers'] = (
    accidents_df.loc[missing_passengers, 'Fatalities'] -
    accidents_df.loc[missing_passengers, 'Fatalities Crew']
)

accidents_df.loc[missing_crew, 'Fatalities Crew'] = (
    accidents_df.loc[missing_crew, 'Fatalities'] -
    accidents_df.loc[missing_crew, 'Fatalities Passangers']
)
```

```python
print(accidents_df['Fatalities Passangers'].isna().sum())
print(accidents_df['Fatalities Crew'].isna().sum())
```

```python
accidents_df['Sum of Aboard'] = (
    accidents_df['Aboard Passangers'] +
    accidents_df['Aboard Crew']
)
```

```python
diff = accidents_df['Aboard'].compare(
    accidents_df['Sum of Aboard']
)

print(diff)
```

```python
accidents_df['Aboard'] = accidents_df['Aboard'].fillna(
    accidents_df['Sum of Aboard']
)
```

```python
print(accidents_df['Aboard'].isna().sum())
```

```python
missing_passengers = (
    accidents_df['Aboard Passangers'].isna() &
    accidents_df['Aboard'].notna() &
    accidents_df['Aboard Crew'].notna()
)

missing_crew = (
    accidents_df['Aboard Crew'].isna() &
    accidents_df['Aboard'].notna() &
    accidents_df['Aboard Passangers'].notna()
)

print("Recoverable passenger values:", missing_passengers.sum())
print("Recoverable crew values:", missing_crew.sum())
```

```python
accidents_df.loc[missing_passengers, 'Aboard Passangers'] = (
    accidents_df.loc[missing_passengers, 'Aboard'] -
    accidents_df.loc[missing_passengers, 'Aboard Crew']
)

accidents_df.loc[missing_crew, 'Aboard Crew'] = (
    accidents_df.loc[missing_crew, 'Aboard'] -
    accidents_df.loc[missing_crew, 'Aboard Passangers']
)
```

```python
accidents_df.fillna({
    'Operator': 'Unknown',
    'Location': 'Unknown',
    'AC Type': 'Unknown',
    'Route': 'Unknown',
    'Registration': 'Unknown'
}, inplace=True)
```

```python
accidents_df['Survival Rate'].agg(['mean', 'median', 'std'])
accidents_df['Survival Rate'] = (
    accidents_df['Survived'] /
    accidents_df['Aboard'].where(accidents_df['Aboard'] > 0)
) * 100
```

```python
accidents_df['decades'] = (accidents_df['Date'].dt.year // 10) * 10
```

```python
# Exercise 4
from scipy import stats

groups = [
    group['Fatalities'].dropna().values
    for decade, group in accidents_df.groupby('decades')
]
```

```python
f_statistic, p_value = stats.f_oneway(*groups)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```

```python
# Hypothesis: The average number of fatalaties per crash is hte same across all decades
# The p value does not support this hypothesys
```
