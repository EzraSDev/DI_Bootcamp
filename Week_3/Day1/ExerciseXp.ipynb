```python
Q1 = train_df["Fare"].quantile(0.25)
Q3 = train_df["Fare"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower boundary:", lower_bound)
print("Upper boundary:", upper_bound)
```

```python
outliers = train_df[train_df["Fare"] > upper_bound]

print("Number of outliers:", len(outliers))

outliers["Fare"].sort_values(ascending=False)

outliers["Pclass"].value_counts()
```

```python
fare_98 = train_df["Fare"].quantile(0.98)

print(fare_98)
```

```python
train_df["Fare"] = train_df["Fare"].clip(upper=fare_98)
print("Maximum fare:", train_df["Fare"].max())
print("Number of rows:", train_df.shape[0])
```

```python
Q1 = train_df["Age"].quantile(0.25)
Q3 = train_df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
age_outliers = train_df[
    (train_df["Age"] < lower_bound) |
    (train_df["Age"] > upper_bound)
]

print("Number of Age outliers:", len(age_outliers))
print(age_outliers["Age"].sort_values())
```
