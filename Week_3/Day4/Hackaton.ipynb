```python
def budget_menu():
    print("Select your vacation length:")
    print("1. 3 days")
    print("2. 7 days")
    print("3. 14 days")

    while True:
        choice = input("Enter your choice (1-3): ")

        if choice in ["1", "2", "3"]:
            days = {"1": 3, "2": 7, "3": 14}[choice]
            break

        print("Invalid choice. Please try again.")

    while True:
        try:
            budget = float(input(
                "Enter your total vacation budget in USD (0 for no limit): "
            ))

            if budget >= 0:
                break

            print("Budget cannot be negative.")

        except ValueError:
            print("Please enter a valid number.")

    return days, budget
```

```python
def filter_budget(df, days, budget):
    cost_column = f"{days} Days"

    if budget == 0:
        return df.copy()

    refined_df = df[df[cost_column] <= budget].copy()

    return refined_df
```

```python
def predict_occupancy(country, year, month):
    data = pd.DataFrame({
        "geo_month": [f"{country}_{month}"],
        "year": [year]
    })

    occupancy = float(model.predict(data)[0])
    occupancy = max(0.0, min(100.0, occupancy))

    return round(occupancy, 2), classify_crowd(occupancy)
```

```python
import matplotlib.pyplot as plt

def plot_monthly_occupancy(df, country):
    country_df = df[df["geo"] == country]

    plt.plot(country_df["month"], country_df["mean"])
    plt.xlabel("Month")
    plt.ylabel("Average Occupancy (%)")
    plt.title(f"Average Monthly Occupancy - {country}")
    plt.show()
```

```python
def add_occupancy(df, year, month):
    df = df.copy()

    predictions = df["country"].apply(
        lambda country: predict_occupancy(country, year, month)
    )

    df["predicted_occupancy"] = predictions.apply(lambda x: x[0])
    df["crowd_level"] = predictions.apply(lambda x: x[1])

    return df
```

```python
def classify_crowd(occupancy):
    if occupancy < 20:
        return "Very Low"
    elif occupancy < 40:
        return "Low"
    elif occupancy < 60:
        return "Moderate"
    elif occupancy < 80:
        return "High"
    else:
        return "Very High"
```

```python
def filter_acttivities_df(df):
    activities = {
        1: "sightseeing",
        2: "beaches",
        3: "nightlife",
        4: "nature",
        5: "hiking",
        6: "skiing",
        7: "culture",
        8: "resorts_spa",
        9: "shopping"
    }

    selected_activities = []

    while True:
        choice = input("""
What activities are you interested in?

1. Sightseeing & Landmarks
2. Beaches & Relaxation
3. Nightlife & Entertainment
4. Nature & Scenery
5. Hiking & Trekking
6. Skiing & Winter Sports
7. Culture, Art & Museums
8. Resorts, Spas & Wellness
9. Shopping

0. Done selecting activities

Enter your choice: """)

        if not choice.isdigit():
            print("Please enter a valid number.")
            continue

        choice = int(choice)

        if choice == 0:
            break

        elif choice not in activities:
            print("Invalid choice. Please select a number between 0 and 9.")

        elif activities[choice] in selected_activities:
            print("You've already selected that activity!")

        else:
            selected_activities.append(activities[choice])
            print(f"Added: {activities[choice]}")

    if selected_activities:
        filtered_df = df[df[selected_activities].any(axis=1)].copy()
    else:
        filtered_df = df.copy()

    return filtered_df
```

```python
def filter_month(df):

    while True:
        choice = input("""
When would you like to travel?

1. January
2. February
3. March
4. April
5. May
6. June
7. July
8. August
9. September
10. October
11. November
12. December

0. No preferred month

Enter your choice: """)

        if not choice.isdigit():
            print("Please enter a valid number.")
            continue

        choice = int(choice)

        if choice < 0 or choice > 12:
            print("Invalid choice. Please select a number between 0 and 12.")
            continue

        break

    if choice == 0:
        return df.copy(), None

    selected_month = months[choice - 1]
    score_column = f"{selected_month}_visit_score"

    filtered_df = df.sort_values(
        by=score_column,
        ascending=False
    ).copy()

    return filtered_df, choice
```

```python
def generate_recomendation(df):
    refined_df = filter_acttivities_df(df)
    refined_df, month = filter_month(refined_df)

    year = int(input("What year are you planning to travel? "))

    days, budget = budget_menu()
    refined_df = filter_budget(refined_df, days, budget)

    if refined_df.empty:
        print("Sorry, no destinations match your preferences.")
        return refined_df

    if month is not None:
        refined_df = add_occupancy(refined_df, year, month)
    else:
        print("No month selected, so occupancy cannot be predicted.")

    return refined_df
```
